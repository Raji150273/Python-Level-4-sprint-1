```python
from typing import Dict, Any, List, Optional, Tuple

# ----- Config -----
allowed_regions = {"North", "South", "East", "West"}


# ----- Parsing & Validation -----
def parse_customer(line: str) -> Tuple[Optional[Dict[str, Any]], Optional[str]]:
    parts = line.split()
    if len(parts) != 4:
        return None, "Input must contain exactly 4 values: <name> <age> <spending> <region>"

    name, age_str, spend_str, region = parts

    # age
    try:
        age = int(age_str)
        if not (18 <= age <= 100):
            return None, "Age must be between 18 and 100."
    except ValueError:
        return None, "Age must be a valid integer."

    # spending
    try:
        spending = float(spend_str)
        if spending <= 0:
            return None, "Monthly spending must be a positive number."
    except ValueError:
        return None, "Monthly spending must be a valid number."

    # region (case-insensitive, stored normalized)
    region_normalized = region.capitalize()
    if region_normalized not in allowed_regions:
        return None, f"Region must be one of: {', '.join(allowed_regions)}."

    customer = {
        "name": name,
        "age": age,
        "spend": spending,
        "region": region_normalized,
    }
    return customer, None


# ----- Input helpers -----
def get_ideal_customer() -> Dict[str, Any]:
    while True:
        line = input("Enter ideal customer (<name> <age> <spending> <region>): ").strip()
        customer, error = parse_customer(line)
        if error:
            print("Error:", error)
            continue
        return customer


def get_other_customers() -> List[Dict[str, Any]]:
    customers: List[Dict[str, Any]] = []

    while len(customers) < 16:
        line = input(
            f"Enter customer {len(customers)+1} (<name> <age> <spending> <region>) "
            "or 'done': "
        ).strip()

        if line.lower() == "done":
            if len(customers) == 0:
                print("You must enter at least one customer before finishing.")
                continue
            break

        customer, error = parse_customer(line)
        if error:
            print("Error:", error)
            continue

        customers.append(customer)
        print(f"Customer '{customer['name']}' added. ({len(customers)}/16)")

    return customers


# ----- Matching logic -----
def is_customer_match(ideal: Dict[str, Any], customer: Dict[str, Any]) -> bool:
    age_match = abs(customer["age"] - ideal["age"]) <= 5
    spend_margin = ideal["spend"] * 0.20
    spend_match = abs(customer["spend"] - ideal["spend"]) <= spend_margin
    region_match = customer["region"] == ideal["region"]
    return age_match and spend_match and region_match


# ----- Classification & report -----
def classify_and_report(ideal: Dict[str, Any], others: List[Dict[str, Any]]) -> None:
    matched: List[Dict[str, Any]] = []
    not_matched: List[Dict[str, Any]] = []

    for customer in others:
        if is_customer_match(ideal, customer):
            matched.append(customer)
        else:
            not_matched.append(customer)

    print("\n===== CUSTOMER MATCH REPORT =====\n")

    print("Ideal customer:")
    print(f" - {ideal['name']} (Age {ideal['age']}, Spend {ideal['spend']}, Region {ideal['region']})\n")

    print("Customers who MATCHED the ideal profile:")
    if matched:
        for c in matched:
            print(f" - {c['name']} (Age {c['age']}, Spend {c['spend']}, Region {c['region']})")
    else:
        print("   None")

    print("\nCustomers who did NOT match the ideal profile:")
    if not_matched:
        for c in not_matched:
            print(f" - {c['name']} (Age {c['age']}, Spend {c['spend']}, Region {c['region']})")
    else:
        print("   None")

    print("\n=================================\n")


# ----- Main entry point -----
def main() -> None:
    """
    Runs the Customer Segmentation Checker.
    """
    print("Welcome to the Customer Segmentation Checker")
    print("First, define the ideal customer.")
    ideal = get_ideal_customer()

    print("\nNow, enter other customers (type 'done' to finish).")
    others = get_other_customers()

    print("\nAnalysing matches...\n")
    classify_and_report(ideal, others)


if __name__ == "__main__":
    main()
```
